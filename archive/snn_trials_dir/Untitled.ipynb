```python
import pandas as pd
```

```python
df1 = pd.read_csv("/Users/kruthikhulisandra/Desktop/SpotifyProject/Datasets/merged_songs.csv")
df2 = pd.read_csv("/Users/kruthikhulisandra/Desktop/SpotifyProject/Datasets/dfSCleaned.csv")
```

```python
df = pd.merge(df1, df2, left_on='Track ID', right_on='spotifyID', how='inner')
```

```python
df
```

```python
df = df.drop(columns=['spotifyID'])
```

```python
df.columns
```

```python

# 1️⃣ Normalize column names
df.columns = (
    df.columns
      .str.strip()                 # remove leading/trailing spaces
      .str.lower()                 # make lowercase
      .str.replace(r'\s+', '_', regex=True)  # replace spaces with underscores
      .str.replace(r'_x$|_y$', '', regex=True)  # remove merge suffixes
)

# 2️⃣ Drop duplicate column names (now case-insensitive)
df = df.loc[:, ~df.columns.duplicated()]

# 3️⃣ Move 'date' and 'time' to positions 2 and 3
cols = df.columns.tolist()
cols.insert(1, cols.pop(cols.index('date')))
cols.insert(2, cols.pop(cols.index('time')))
df = df[cols]
```

```python
keep_cols = [
    # Identification & Time
    'track_id', 'date', 'time',
    
    # Playback Session Details
    'conn_country', 'platform', 'ms_played', 'skipped',
    'reason_start', 'reason_end', 'shuffle', 'offline', 'incognito_mode',
    
    # Song Metadata
    'track_name', 'artist_name', 'album_name', 'genres', 'release_date',
    'record_label', 'added_by', 'added_at',
    
    # Audio Features
    'duration_(ms)', 'popularity',
    'danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness',
    'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo',
    'time_signature'
]
```

```python
df = df[keep_cols]
```
