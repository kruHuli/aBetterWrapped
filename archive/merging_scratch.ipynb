```python
import pandas as pd
import glob

# Get all CSV file names in the directory
csv_files = ["sad_beats.csv", "emotional_dance.csv", "stutter_house.csv", "varsha__kruthik.csv", "liked_songs.csv"]

# Read and concatenate all CSV files
df_list = [pd.read_csv(file) for file in csv_files]
merged_df = pd.concat(df_list, ignore_index=True)

# Remove duplicates based on "Track ID"
merged_df = merged_df.drop_duplicates(subset=["Track ID"])

# Save the merged CSV
merged_df.to_csv("merged_songs.csv", index=False)

print("CSV files merged successfully into 'merged_songs.csv'!")
```

```python
import pandas as pd

# Load the CSV files
dfAN24 = pd.read_csv('dfAN24.csv')
merged_songs = pd.read_csv('merged_songs.csv')

# Rename columns to match for merging
# Assuming 'spotifyID' in dfAN24 corresponds to 'Track ID' in merged_songs
dfAN24 = dfAN24.rename(columns={'spotifyID': 'Track ID'})

# Merge the dataframes on 'Track ID'
result = pd.merge(
    dfAN24, 
    merged_songs, 
    on='Track ID', 
    how='inner',  # Use 'inner' to keep only matching rows, use 'outer', 'left' or 'right' for different join types
    suffixes=('_dfAN24', '_merged')  # Add suffixes to distinguish columns with the same name
)

# If there are duplicate column names (like 'Energy'), you might want to handle them
# You can either keep both columns with suffixes or drop one of them
# Here we're keeping both with suffixes applied automatically

# Save the result to a new CSV file
result.to_csv('merged_result.csv', index=False)

print(f"Merged CSV created with {len(result)} rows")
```
